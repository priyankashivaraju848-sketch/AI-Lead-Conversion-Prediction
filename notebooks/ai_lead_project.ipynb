```python
X = df_selected.drop('Converted', axis=1)
y = df_selected['Converted']

print("Input features:", X.shape)
print("Target variable:", y.shape)
```

```python
categorical_cols = X.select_dtypes(
    include=['object']
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Categorical columns:", categorical_cols)
print("Numerical columns:", numerical_cols)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Numerical preprocessing
numerical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore',
        drop='first'
    ))
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ('num', numerical_pipeline, numerical_cols),
    ('cat', categorical_pipeline, categorical_cols)
])

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply same transformations to testing data
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from scipy import sparse
import numpy as np

print("Training shape:", X_train_processed.shape)
print("Testing shape:", X_test_processed.shape)

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

# Check missing values in training data
if sparse.issparse(X_train_processed):
    print("Missing values in training:",
          np.isnan(X_train_processed.data).sum())
else:
    print("Missing values in training:",
          np.isnan(X_train_processed).sum())

# Check missing values in testing data
if sparse.issparse(X_test_processed):
    print("Missing values in testing:",
          np.isnan(X_test_processed.data).sum())
else:
    print("Missing values in testing:",
          np.isnan(X_test_processed).sum())
```

```python
import joblib
import pandas as pd

# Save preprocessing pipeline
joblib.dump(preprocessor, 'preprocessor.pkl')

# Save training and testing datasets
joblib.dump(X_train_processed, 'X_train_processed.pkl')
joblib.dump(X_test_processed, 'X_test_processed.pkl')

# Save target variables
joblib.dump(y_train, 'y_train.pkl')
joblib.dump(y_test, 'y_test.pkl')

# Save cleaned dataset
df_selected.to_csv('cleaned_leads.csv', index=False)

# Save feature names after preprocessing
feature_names = preprocessor.get_feature_names_out()

pd.DataFrame({
    'Feature Name': feature_names
}).to_csv('processed_feature_names.csv', index=False)

print("All files saved successfully!")
```

```python
def generate_followup_strategy(lead_details, probability, category):

    prompt = f"""
    You are an AI sales follow-up assistant.

    Lead Details: {lead_details}
    Conversion Probability: {probability * 100:.2f}%
    Lead Category: {category}

    Generate:

    1. Follow-up objective
    2. Communication channel
    3. Suggested timing
    4. Personalized follow-up message
    5. Next recommended action

    Use only the available lead information.
    Do not invent customer details.
    """

    response = ai_client.chat.completions.create(
        model="openrouter/free",
        messages=[
            {
                "role": "system",
                "content": "You are an expert sales follow-up assistant."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_tokens=500
    )

    message = response.choices[0].message

    result = message.content

    if not result and message.reasoning:
        result = message.reasoning

    if not result:
        raise ValueError("No text returned by the AI model.")

    return result
```

```python
import time

def ai_sales_assistant(question, lead_details, probability, category):

    prompt = f"""
    You are an expert AI sales assistant.

    Lead Details: {lead_details}
    Conversion Probability: {probability * 100:.2f}%
    Lead Category: {category}

    Salesperson's Question: {question}

    Answer the question using the supplied lead information.

    Give practical and clear sales suggestions.
    Do not invent customer information.
    Do not guarantee conversion.
    """

    for attempt in range(3):

        try:
            response = ai_client.chat.completions.create(
                model="openrouter/free",
                messages=[
                    {
                        "role": "system",
                        "content": "You are an intelligent AI sales assistant."
                    },
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                max_tokens=500
            )

            result = response.choices[0].message.content

            if result and result.strip():
                return result

            print("Empty response. Retrying...")

        except Exception as e:
            print("Attempt", attempt + 1, "failed:", str(e))

        if attempt < 2:
            time.sleep(2)

    return "AI service is temporarily unavailable. Please try again."
```
